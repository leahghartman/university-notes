```python
import numpy as np
import matplotlib.pyplot as plt

mu = np.linspace(-10,0,400)

z1=[]
z2=[]

for m in mu:
    a = 1.5 - m
    b = -2
    c = 0.5
    r = np.roots([a,b,c])
    z1.append(abs(r[0]))
    z2.append(abs(r[1]))

plt.plot(mu,z1,label=r'$|\zeta_1|$')
plt.plot(mu,z2,label=r'$|\zeta_2|$')
plt.xlabel('hλ', fontsize=13)
plt.ylabel(r'$|\zeta|$', fontsize=13)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Parameters
sigma = 10
b = 8/3
r = 28

# Lorenz system
def lorenz(t, y):
    y1, y2, y3 = y
    dy1 = sigma * (y2 - y1)
    dy2 = r*y1 - y2 - y1*y3
    dy3 = y1*y2 - b*y3
    return [dy1, dy2, dy3]

# Initial conditions
y0 = [0, 1, 0]

# Time span
t_span = (0, 100)
t_eval = np.linspace(0, 100, 20000)

# Solve ODE (RK45 similar to Matlab ode45)
sol = solve_ivp(lorenz, t_span, y0, method='RK45', t_eval=t_eval)

# Extract solution
y1 = sol.y[0]
y3 = sol.y[2]

# Plot projection onto y1-y3 plane
plt.figure()
plt.plot(y1, y3, linewidth=0.5)
plt.xlabel(r"$y_1$", fontsize=13)
plt.ylabel(r"$y_3$", fontsize=13)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
h = 0.05
x = np.arange(0, 1 + h, h)
Nx = len(x)
k_values = [0.0013, 0.0012]

# Initial conditions
def initial_a(x):
    return 1 - 2*np.abs(x - 0.5)

def initial_b(x):
    return np.sin(np.pi * x)

def solve_heat(k, initial_func):
    r = k / h**2
    v = initial_func(x)
    snapshots = {}
    times = [0, 1, 25, 50]

    for n in range(51):
        if n in times:
            snapshots[n] = v.copy()
        v_new = v.copy()
        for i in range(1, Nx-1):
            v_new[i] = v[i] + r * (v[i+1] - 2*v[i] + v[i-1])
        v_new[0] = 0
        v_new[-1] = 0
        v = v_new

    return snapshots

# 2x2 subplot grid: rows = initial condition, cols = k value
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for col, k in enumerate(k_values):
    snapshots_a = solve_heat(k, initial_a)
    snapshots_b = solve_heat(k, initial_b)

    for n, v in snapshots_a.items():
        axes[0, col].plot(x, v, label=rf"$t = {n}k$")
    axes[0, col].set_title(f"Initial condition (a), k = {k}", fontsize=14)
    axes[0, col].set_xlabel(r"$x$", fontsize=13)
    axes[0, col].set_ylabel(r"$v(x,t)$", fontsize=13)
    axes[0, col].legend()

    for n, v in snapshots_b.items():
        axes[1, col].plot(x, v, label=rf"$t = {n}k$")
    axes[1, col].set_title(f"Initial condition (b), k = {k}", fontsize=14)
    axes[1, col].set_xlabel(r"$x$", fontsize=13)
    axes[1, col].set_ylabel(r"$v(x,t)$", fontsize=13)
    axes[1, col].legend()

plt.tight_layout()
plt.show()
```
