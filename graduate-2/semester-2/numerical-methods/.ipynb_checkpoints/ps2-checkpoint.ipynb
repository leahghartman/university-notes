```python
import numpy as np
import matplotlib.pyplot as plt
```

# PROBLEM 3:

```python
h_vals = [0.11, 0.10, 0.09]

def forward_euler(n,h):
  return 1.1*(1-2*h)**n - 0.1*(1-20*h)**n

def backward_euler(n,h):
  return 1.1*(1+2*h)**(-n) - 0.1*(1+20*h)**(-n)

def exact(t):
  return 1.1*np.exp(-2*t) - 0.1*np.exp(-20*t)

# compute u_FE for n = 100 steps
plt.figure(figsize=(12, 12))
for i, h in enumerate(h_vals):
  n = int(2/h)
  t = np.arange(n + 1) * h

  u_FE = np.zeros(n + 1)
  u_BE = np.zeros(n + 1)
  u_EX = np.zeros(n + 1)

  for N in range(n+1):
    u_FE[N] = forward_euler(N,h)
    u_BE[N] = backward_euler(N,h)
    u_EX[N] = exact(t[N])

  # ----- Forward Euler (left column)
  plt.subplot(3, 2, 2*i + 1)
  plt.plot(t, u_EX, 'k-', linewidth=2, label='Exact')
  plt.plot(t, u_FE, 'ro--', label='Forward Euler')
  plt.title(rf'Forward Euler, $h = ${h}', fontsize=13)
  plt.xlabel(r'$t$', fontsize=12)
  plt.ylabel(r'$y_1(t)$', fontsize=12)
  plt.grid(True)

  # ----- Backward Euler (right column)
  plt.subplot(3, 2, 2*i + 2)
  plt.plot(t, u_EX, 'k-', linewidth=2, label='Exact')
  plt.plot(t, u_BE, 'bs--', label='Backward Euler')
  plt.title(rf'Backward Euler, $h = ${h}', fontsize=13)
  plt.xlabel(r'$t$', fontsize=12)
  plt.ylabel(r'$y_1(t)$', fontsize=12)
  plt.grid(True)

plt.tight_layout()
plt.subplots_adjust(hspace=0.4, wspace=0.3)
plt.show()
```

# PROBLEM 4:

```python
# set the step size and the number of steps
h = 4*np.pi/100
n = 100

# eigenvectors
v1 = np.array([1,1j])
v2 = np.array([1,-1j])

# compute u_FE for n = 100 steps
u_FE = np.zeros((n+1,2), dtype=complex)
for N in range(n+1):
  u_FE[N] = 0.5*(1+1j*h)**N * v1 + 0.5*(1-1j*h)**N * v2

# compute u_BE for n = 100 steps
u_BE = np.zeros((n+1,2), dtype=complex)
for N in range(n+1):
  u_BE[N] = 0.5*(1-1j*h)**(-N) * v1 + 0.5*(1+1j*h)**(-N) * v2

# --- RK4 ---
# define the A matrix and the u0 matrix
A = np.array([[0,1], [-1,0]])
u0 = np.array([1.0,0.0])

# define a function that steps through rk4
def rk4(u,A):
  k1 = A @ u
  k2 = A @ (u + 0.5*h*k1)
  k3 = A @ (u + 0.5*h*k2)
  k4 = A @ (u + h*k3)

  return u + h/6*(k1 + 2*k2 + 2*k3 + k4)

u_RK = np.zeros((n+1,2))
u_RK[0] = u0
for N in range(n):
  u_RK[N+1] = rk4(u_RK[N],A)
  

u_FE = u_FE.real
u_BE = u_BE.real
```

```python
t = np.arange(n+1)*h

# define the exact solution
y_exact = np.zeros((n+1,2))
y_exact[:,0] = np.cos(t)
y_exact[:,1] = -np.sin(t)
```

```python
plt.figure(figsize=(6,6))
plt.plot(y_exact[:,0], y_exact[:,1], color='black', label='Exact Solution')
plt.plot(u_FE[:,0], u_FE[:,1], 'r--', label='Forward Euler')
plt.plot(u_BE[:,0], u_BE[:,1], 'b--', label='Backward Euler')
plt.plot(u_RK[:,0], u_RK[:,1], 'g--', label='RK4')

plt.xlabel(r'$y_1(t)$', fontsize=13)
plt.ylabel(r'$y_2(t)$', fontsize=13)
plt.legend()
plt.axis('equal');
plt.show()
```
