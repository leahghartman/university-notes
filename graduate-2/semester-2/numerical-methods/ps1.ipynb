```python
import numpy as np
import matplotlib.pyplot as plt

# function and exact derivative
def f(x):
  return np.exp(x)

def df_exact(x):
  return np.exp(x)

x = 1.0
exact_value = df_exact(x)

# step sizes
j = np.arange(1, 66)
h = 2.0**(-(j-1))

# arrays for errors
error_forward = np.zeros_like(h)
error_centered = np.zeros_like(h)

for k in range(len(h)):
    # Forward difference
    D_plus = (f(x + h[k]) - f(x)) / h[k]
    
    # Centered difference
    D_zero = (f(x + h[k]) - f(x - h[k])) / (2*h[k])
    
    # Errors
    error_forward[k] = abs(D_plus - exact_value)
    error_centered[k] = abs(D_zero - exact_value)

# log-log plot
plt.loglog(h, error_forward, '.-', label='Forward difference')
plt.loglog(h, error_centered, 's-', ms=3, label='Centered difference')

plt.xlabel(r'$h$')
plt.ylabel(r'$|f\'(x) - Df(x)|$')
plt.legend()
plt.grid(True, which='both')

plt.show()
```
