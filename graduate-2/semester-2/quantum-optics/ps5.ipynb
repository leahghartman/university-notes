```python
import numpy as np
import matplotlib.pyplot as plt
from math import factorial
```

# Problem 4 (Gerry-Knight 4.4)

We found that, when the field is initially in a coherent state, $\langle \hat{d} \rangle$ looks like:
$$
\begin{equation*}
    \langle \hat{d} \rangle = -2d \Im(\alpha) e^{-|\alpha|^2} \sum_{n=0}^{\infty} \frac{|\alpha|^{2n}}{n!\sqrt{n+1}}\sin(\lambda t\sqrt{n+1})\cos(\lambda t\sqrt{n+2}),
\end{equation*}
$$
which we can plot using the below code.

```python
# --- Choose some initial conditions ---
alpha = 3.0 + 0.5j       # Choose the coherent state amplitude
lam = 1.0                # Sets the Rabi frequency scale
d = 1.0                  # Sets the overall amplitude of <d>
nmax = 50                # Truncate the sum to n = 50

# Create a time array and an array to hold <d> data
t = np.linspace(0,20,2000)
dipole = np.zeros_like(t)

# --- For loop to cycle through values of n ---
for n in range(nmax):
    weight = (abs(alpha)**(2*n)) / (factorial(n)*np.sqrt(n+1))
    dipole += weight * np.sin(lam*np.sqrt(n+1)*t) * np.cos(lam*np.sqrt(n+2)*t)

# --- Can't forget to multiply the prefactor by the sum ---
prefactor = -2*d*np.imag(alpha)*np.exp(-abs(alpha)**2)
dipole = prefactor * dipole

# --- Plot everything ---
plt.plot(t,dipole)
plt.xlabel(r"$t$", fontsize=13)
plt.ylabel(r"$\langle \hat{d} \rangle$", fontsize=13)
plt.show()
```

# Problem 6 (Berman-Malinovsky 15.6)

We found that $P_2(t) = \frac{1}{2}\sin^2(gt) + \frac{1}{2}\sin^2(\sqrt{2}gt)$, so we plot this below.

```python
# -- Sets the Rabi frequency scale; choose something nice and easy like 1.0 ---
g = 1.0

# --- Create a time array ---
t = np.linspace(0,20,2000)

# --- Write the P_2(t) equation and get values for times in time array ---
P2 = 0.5*(np.sin(g*t)**2 + np.sin(np.sqrt(2)*g*t)**2)

# --- Plot everything ---
plt.figure()
plt.plot(t,P2)
plt.xlabel(r"$t$", fontsize=13)
plt.ylabel(r"$P_2(t)$", fontsize=13)
plt.show()
```
